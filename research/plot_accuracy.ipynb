# Training-accuracy figures

Bar charts of train / validation accuracy for the two EfficientNet models (original; synthetic→original). Reads the saved `history_*.json` files and writes the figures to `artifacts/`.

```python
import json

import numpy as np
from matplotlib import pyplot as plt
```

```python
with open('../models/efficientnet_v2_b3/original_data/history_original_data_epoch_6.json') as f:
    history = json.load(f)

width = 0.3
plt.figure(figsize=(10, 5))
plt.bar(np.arange(6) + 1 - width / 2, history['accuracy'], width, alpha=0.5)
plt.bar(np.arange(6) + 1 + width / 2, history['val_accuracy'], width, alpha=0.5)
plt.title('Classification accuracy — EfficientNetV2-B3 (original data)', fontsize=16)
plt.ylabel('Accuracy', fontsize=16)
plt.xlabel('Epoch', fontsize=16)
plt.legend(['training accuracy', 'validation accuracy'], loc='lower right', fontsize=12)
plt.ylim([0.975, 1.0])
plt.savefig('artifacts/orig_accuracy.png')
```

```python
with open('../models/efficientnet_v2_b3/combined_data/history_synthetic_data_epoch_6.json') as f:
    history1 = json.load(f)

with open('../models/efficientnet_v2_b3/combined_data/history_original_data_epoch_3.json') as f:
    history2 = json.load(f)

plt.figure(figsize=(15, 5))
plt.bar(np.arange(6) + 1 - width / 2, history1['accuracy'], width, alpha=0.5, color='C2')
plt.bar(np.arange(6) + 1 + width / 2, history1['val_accuracy'], width, alpha=0.5, color='C3')
plt.bar(np.arange(3) + 7 - width / 2, history2['accuracy'], width, alpha=0.5, color='C0')
plt.bar(np.arange(3) + 7 + width / 2, history2['val_accuracy'], width, alpha=0.5, color='C1')
plt.title('Classification accuracy — EfficientNetV2-B3 (synthetic → original)', fontsize=16)
plt.ylabel('Accuracy', fontsize=16)
plt.xlabel('Epoch', fontsize=16)
plt.legend(['synthetic — training', 'synthetic — validation',
            'original — training', 'original — validation'], loc='lower left', fontsize=16)
plt.ylim([0.965, 1.0])
plt.savefig('artifacts/syn_orig_accuracy.png')
```
